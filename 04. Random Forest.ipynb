```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
```

```python
dataset
```

```python
dataset.columns
```

```python
Independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
```

```python
Independent
```

```python
dependent=dataset[['charges']]
```

```python
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(Independent, dependent, test_size=0.30, random_state=0)
```

```python
from sklearn.ensemble import RandomForestRegressor
regressor=RandomForestRegressor(criterion='absolute_error',max_features='sqrt',n_estimators=100)
regressor=regressor.fit(x_train,y_train)
```

```python
y_pred=regressor.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test, y_pred)
```

```python
r_score
```

```python
import pickle
filename="04.finalized_model_Random_Forest.sav"
```

```python
pickle.dump(regressor,open(filename,'wb'))
```

```python
load_model=pickle.load(open("04.finalized_model_Random_Forest.sav",'rb'))
result=load_model.predict([[1,2,3,4,5]])
```
